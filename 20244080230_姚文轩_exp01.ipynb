## 实验1：基本运算、数据类型与字符串

**思路**

- `/` 是真除法，结果一定是 float；`//` 向下取整，`%` 取余，`**` 求幂。
- 用 `type()` 查看每个值的类型。
- 字符串与数值之间的转换很常用，转换失败会抛 `ValueError`，用 `try / except` 捕获并说明原因。
- 字符串是**不可变**对象，"修改某个字符"只能靠切片拼出新字符串。

```python
print("8 / 3  =", 8 / 3, " 类型:", type(8 / 3).__name__)
print("8 // 3 =", 8 // 3, " 类型:", type(8 // 3).__name__)
print("8 % 3  =", 8 % 3, " 类型:", type(8 % 3).__name__)
print("2 ** 7 =", 2 ** 7, " 类型:", type(2 ** 7).__name__)
```

四个运算符的含义：

| 运算符 | 含义 | 说明 |
| --- | --- | --- |
| `/` | 真除法 | 8 / 3 = 2.666…，**结果一定是 float** |
| `//` | 向下取整除法 | 8 // 3 = 2，结果类型跟操作数有关 |
| `%` | 取模（求余数） | 8 % 3 = 2；结果的符号跟随除数 |
| `**` | 幂运算 | 2 ** 7 = 128，2 的 7 次方 |

```python
samples = {"整数": 42, "浮点数": 3.14, "字符串": "Python", "布尔值": True}
for label, value in samples.items():
    print(f"{label:<4}: {value!r:<10} 类型 = {type(value).__name__}")
```

```python
print('int("123")  =', int("123"), "  类型:", type(int("123")).__name__)
print('float("3.14")  =', float("3.14"), "  类型:", type(float("3.14")).__name__)
print("int(3.9)    =", int(3.9), "  类型:", type(int(3.9)).__name__)

try:
    int("goat")
except ValueError as exc:
    print('int("goat") 抛出:', type(exc).__name__, "-", exc)
```

转换结果说明：

- `int("123")`：字符串 `"123"` 是合法的整数写法，转成整数 123。
- `float("3.14")`：字符串转成浮点数 3.14。
- `int(3.9)`：**直接截断小数部分**（不是四舍五入），得到 3。
- `int("goat")`：抛出 `ValueError`。原因是 `"goat"` 不构成一个合法的十进制整数，
  `int()` 无法把它解释成数值（异常信息里会说明是 invalid literal for int()）。

## 实验2：函数、条件、循环与递归

**思路**

- `score_level()` 用 `if / elif` 链按区间返回等级，越界时主动抛 `ValueError`。
- 1~100 求和分别用 `for` 和 `while` 实现，比较两种写法的结果是否一致。
- 阶乘写递归版和迭代版：递归直接对应定义 `n! = n × (n-1)!`，迭代用循环累乘；`0! = 1` 是递归的终止条件。
- `summary()` 用元组一次返回最小值、最大值、平均值，调用处用元组解包接收。
- `sum_all(*args)` 用可变参数接收任意个参数，无参数时和为 0。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在 0~100 之间，收到: {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for s in [100, 95, 90, 89, 80, 79, 60, 59, 0]:
    print(f"score_level({s:>3}) = {score_level(s)}")

try:
    score_level(105)
except ValueError as exc:
    print("score_level(105) 抛出:", type(exc).__name__, "-", exc)
```

```python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   循环求和 1~100 =", total_for)
print("while 循环求和 1~100 =", total_while)
print("两种实现结果相同:", total_for == total_while)
```

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


for n in [0, 1, 5, 10]:
    r, it = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>2}!  递归 = {r:<10} 迭代 = {it:<10} 一致: {r == it}")
```

```python
def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


lowest, highest, average = summary([85, 58, 92, 76, 60, 89])
print("最小值 =", lowest)
print("最大值 =", highest)
print(f"平均值 = {average:.2f}")
print("summary() 直接返回:", summary([10, 20, 30]))
```

```python
def sum_all(*args):
    total = 0
    for value in args:
        total += value
    return total


print("无参数        : sum_all()           =", sum_all())
print("单个参数      : sum_all(5)          =", sum_all(5))
print("多个参数      : sum_all(1, 2, 3, 4) =", sum_all(1, 2, 3, 4))
print("接收到的参数  :", sum_all(*[10, 20, 30]), "（用 * 解包列表传入）")
```

## 实验3：列表、字典、元组与基础统计

**思路**

- 用 `zip()` 把姓名和成绩配对成字典，`items()` 遍历键值对，`enumerate()` 同时拿到序号和元素。
- 列表推导式负责"筛选"（及格学生）和"变换"（成绩平方）。
- 统计量直接用 `len / sum / max / min` 计算，再用格式化字符串控制平均分的小数位和及格率的百分数。
- 第 5 题：`alias = scores` 只是**起别名**（同一个对象），`scores.copy()` 才是新列表；为不破坏后续要用的原始成绩，该观察用单独的演示列表。
- 元组是**不可变**的，元素不能重新赋值；`zip()` 在长度不同时以**较短的序列**为准。

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_map = dict(zip(names, scores))
print("姓名到成绩的字典:", score_map)
print()
for name, score in score_map.items():
    print(f"  {name:<6} {score}")
```

```python
for index, (name, score) in enumerate(score_map.items(), start=1):
    print(f"  {index}. {name:<6} {score}")
```

```python
passed_names = [name for name, score in score_map.items() if score >= 60]
squared_scores = [score ** 2 for score in scores]

print("及格学生姓名 :", passed_names)
print("所有成绩平方 :", squared_scores)
```

```python
count = len(scores)
average = sum(scores) / count
highest = max(scores)
lowest = min(scores)
pass_rate = sum(1 for s in scores if s >= 60) / count

print(f"人数   : {count}")
print(f"平均分 : {average:.2f}")
print(f"最高分 : {highest}")
print(f"最低分 : {lowest}")
print(f"及格率 : {pass_rate:.1%}")
```

```python
demo = [10, 20, 30]
alias = demo               # 直接赋值：只是同一个对象的另一个名字
copied = demo.copy()       # copy()：创建一个新的外层列表

alias[0] = 999

print("demo   :", demo)
print("alias  :", alias)
print("copied :", copied)
```

观察结果：改 `alias[0]` 之后，`demo` **也跟着变了**——因为 `alias = demo` 只是给同一个列表
对象起了第二个名字，并没有复制数据。而 `copied` 是 `demo.copy()` 产生的**新列表**，
内容仍是原来的 `[10, 20, 30]`，不受影响。

```python
record = ("Alice", 85)
print("学生记录元组:", record)
try:
    record[1] = 90
except TypeError as exc:
    print("元组元素不能重新赋值:", type(exc).__name__, "-", exc)

print("长度不同的 zip:", list(zip(["a", "b", "c"], [1, 2])))
```

元组和字符串一样是不可变对象，`record[1] = 90` 会抛 `TypeError`。
`zip()` 在长度不同的两个序列上以**较短的那个**为准，多出来的元素被忽略，
所以 `zip(["a","b","c"], [1,2])` 只产生两组配对。

## 实验4：文件读写、异常处理与模块

**思路**

- 用 `os.getcwd()` 查看工作目录，`os.path.join()` 拼出相对路径，`os.makedirs(..., exist_ok=True)` 建 `exp01_data` 目录。
- 用 `with open(..., encoding="utf-8")` 写入并读回 `students.txt`；读取时跳过表头，逐行用字符串方法拆分。
- 校验逻辑抽成 `parse_student()`：成绩转整数、检查字段数量、姓名是否为空、成绩是否在 0~100，无效记录一律抛 `ValueError`。
- 主循环用 `try / except` 捕获，记下行号和原因后继续处理后面的行，一条坏数据不会让整个程序中断。
- `w` 模式会清空原有内容，`a` 模式在末尾追加；再用 `pickle` 保存有效记录并读回验证相等。
- 把 `parse_student()` 存成同级目录的 `student_utils.py` 模块并 `import` 调用；最后演示 `FileNotFoundError` 的捕获。

```python
import os

print("当前工作目录:", os.getcwd())

DATA_DIR = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(DATA_DIR, exist_ok=True)

print("数据目录    :", DATA_DIR)
print("目录是否已存在:", os.path.isdir(DATA_DIR))
```

```python
STUDENTS_TEXT = '''name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
'''

students_path = os.path.join(DATA_DIR, "students.txt")
with open(students_path, "w", encoding="utf-8") as fh:
    fh.write(STUDENTS_TEXT)
print("已写入:", students_path)

with open(students_path, encoding="utf-8") as fh:
    content = fh.read()
print("--- 文件内容 ---")
print(content, end="")
```

```python
with open(students_path, encoding="utf-8") as fh:
    lines = fh.read().splitlines()

header, data_lines = lines[0], lines[1:]
print("总行数 :", len(lines))
print("表头   :", header)
print("字段名 :", header.split(","))
print("数据行 :", len(data_lines), "行")
```

```python
def parse_student(line):
    '''把 "name,score" 形式的一行解析成字典；无效记录抛出 ValueError。'''
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")
    name, raw_score = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是合法的整数: {raw_score!r}") from None
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}


print("合法行:", parse_student("Alice,85"))
for bad in ["Henry,105", "Grace,abc", "Incomplete", ",90"]:
    try:
        parse_student(bad)
    except ValueError as exc:
        print(f"无效行 {bad!r:<14} -> {exc}")
```

```python
valid_records = []
invalid_records = []

with open(students_path, encoding="utf-8") as fh:
    fh.readline()                                  # 跳过表头
    for lineno, line in enumerate(fh, start=2):    # 行号从 2 开始（第 1 行是表头）
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append((lineno, line.strip(), str(exc)))

print("有效记录数:", len(valid_records))
print("无效记录数:", len(invalid_records))
print()
print("--- 有效记录 ---")
for rec in valid_records:
    print("  ", rec)
print("--- 无效记录（行号 / 原文 / 原因）---")
for lineno, raw, reason in invalid_records:
    print(f"  第 {lineno} 行: {raw!r:<14} -> {reason}")
```

```python
scores_valid = [rec["score"] for rec in valid_records]
avg_valid = sum(scores_valid) / len(scores_valid)
pass_rate_valid = sum(1 for s in scores_valid if s >= 60) / len(scores_valid)

report_path = os.path.join(DATA_DIR, "report.txt")
report_text = "\n".join([
    f"有效记录数: {len(valid_records)}",
    f"无效记录数: {len(invalid_records)}",
    f"平均分: {avg_valid:.2f}",
    f"及格率: {pass_rate_valid:.2%}",
])

with open(report_path, "w", encoding="utf-8") as fh:
    fh.write(report_text + "\n")
print("--- w 模式写入后 ---")
with open(report_path, encoding="utf-8") as fh:
    print(fh.read(), end="")

with open(report_path, "a", encoding="utf-8") as fh:
    fh.write("--- 追加的一行（a 模式）---\n")
print("--- a 模式追加后 ---")
with open(report_path, encoding="utf-8") as fh:
    print(fh.read(), end="")
```

`w` 与 `a` 模式的区别：

- **`w`（写入）**：打开时会**清空原有内容**，然后从头写入。所以上面第二次用 `w` 的话，
  之前写的内容会被丢掉。
- **`a`（追加）**：保留原有内容，把新数据写到**文件末尾**。上面用的就是 `a`，
  所以原来的四行统计信息还在，只是后面多了一行。

```python
import pickle

pkl_path = os.path.join(DATA_DIR, "students.pkl")
with open(pkl_path, "wb") as fh:
    pickle.dump(valid_records, fh)
print("已保存:", pkl_path, "（", os.path.getsize(pkl_path), "字节 ）")

with open(pkl_path, "rb") as fh:
    restored_records = pickle.load(fh)

print("恢复的数据 :", restored_records)
print("与原数据相等:", restored_records == valid_records)
```

## 实验5：类、继承与对象复制

**思路**

- `Student` 类在 `__init__` 里初始化 `name` 和 `score`，实现 `is_passed()` 判断是否及格、`__str__()` 决定 `print(对象)` 时显示什么。
- 为实验4 中每条有效记录创建一个 `Student` 对象，输出对象信息和及格状态。
- `GraduateStudent` 继承 `Student`，增加 `research_topic` 属性，用 `super().__init__()` 复用父类初始化，并重写 `__str__()`。
- 对象复制对比"直接赋值 / `copy.copy()` / `copy.deepcopy()`"：区别只体现在**嵌套的可变对象**上；每次比较都从新建的原始对象开始。

```python
class Student:
    '''学生：姓名 + 成绩。'''

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score})"


demo_student = Student("Alice", 85)
print("对象信息  :", demo_student)
print("是否及格  :", demo_student.is_passed())
bob = Student("Bob", 58)
print("不及格示例:", bob, "->", bob.is_passed())
```

```python
students = [Student(rec["name"], rec["score"]) for rec in valid_records]

print(f"共 {len(students)} 个 Student 对象：")
for stu in students:
    status = "及格" if stu.is_passed() else "不及格"
    print(f"  {stu!s:<46} {status}")
```

```python
class GraduateStudent(Student):
    '''研究生：在学生的属性之外多一个研究方向。'''

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 复用父类的初始化
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"research_topic={self.research_topic!r})")


grad = GraduateStudent("Carol", 92, "机器学习")
print("对象信息        :", grad)
print("是否及格        :", grad.is_passed(), "（is_passed 继承自 Student）")
print("研究方向        :", grad.research_topic)
print("isinstance 判断 :", isinstance(grad, Student))
```

```python
import copy


def make_original():
    '''每次都新建一个原始对象，保证三次比较互不影响。'''
    return {"name": "Alice", "scores": [85, 90]}


o1 = make_original()
a1 = o1                       # 直接赋值
a1["scores"].append(100)
print("直接赋值   : 改副本后，原对象 =", o1)

o2 = make_original()
c2 = copy.copy(o2)            # 浅拷贝
c2["scores"].append(100)
print("copy.copy  : 改副本后，原对象 =", o2)

o3 = make_original()
d3 = copy.deepcopy(o3)        # 深拷贝
d3["scores"].append(100)
print("deepcopy   : 改副本后，原对象 =", o3)
```

三种方式的区别：

- **直接赋值** `a1 = o1`：只是同一个对象的另一个名字，**改副本等于改原对象**。
- **`copy.copy()` 浅拷贝**：只复制最外层字典，内层列表仍然是**共享**的，
  所以往副本的列表里 `append` 会同时影响原对象。
- **`copy.deepcopy()` 深拷贝**：递归复制所有层级，副本与原对象**完全独立**，
  改副本不会碰到原对象。

所以只有在对象内部还嵌套着可变对象（如列表、字典）时，浅拷贝和深拷贝才会表现出差别。
