## 1. 检查字符串是否仅包含英文字母 a-z、A-Z 和数字 0-9

**思路**：`^` 匹配开头，`[a-zA-Z0-9]` 匹配字母数字，`+` 至少 1 个字符，`$` 匹配结尾；空字符串返回 False。

```python
def check_alnum(s: str) -> bool:
    pattern = r'^[a-zA-Z0-9]+$'
    return bool(re.fullmatch(pattern, s))

# 测试用例
print(check_alnum("abc123"))   # True
print(check_alnum("Abc678"))   # True
print(check_alnum("abc_123"))  # False
print(check_alnum("hi!123"))   # False
print(check_alnum(""))         # False
```

## 2. 匹配一个 a 后面跟着零个或多个 b 的字符串

**思路**：`a`固定匹配 a，`b*`代表 0 个或多个 b，`*`允许没有 b。

```python
def match_a_zero_b(s: str) -> bool:
    pattern = r'^ab*$'
    return bool(re.fullmatch(pattern, s))

# 测试用例
print(match_a_zero_b("a"))      # True
print(match_a_zero_b("ab"))     # True
print(match_a_zero_b("abbb"))   # True
print(match_a_zero_b("ac"))     # False
print(match_a_zero_b("abbc"))   # False
```

## 3. 匹配一个 a 后面跟着一个或多个 b 的字符串

**思路**：`b+`代表 1 个及以上 b，不能只有单独 a。

```python
def match_a_one_b(s: str) -> bool:
    pattern = r'^ab+$'
    return bool(re.fullmatch(pattern, s))

# 测试用例
print(match_a_one_b("a"))       # False
print(match_a_one_b("ab"))      # True
print(match_a_one_b("abb"))     # True
print(match_a_one_b("abbbc"))   # False
```

## 4. 匹配以 a 开头、以 b 结尾，且中间可以有任意字符的字符串

**思路**：`^a`匹配 a 开头，`.*`匹配中间任意字符，`b$`匹配 b 结尾。

```python
def match_start_a_end_b(s: str) -> bool:
    pattern = r'^a.*b$'
    return bool(re.fullmatch(pattern, s))

# 测试用例
print(match_start_a_end_b("ab"))      # True
print(match_start_a_end_b("a123b"))   # True
print(match_start_a_end_b("aXyZb"))   # True
print(match_start_a_end_b("a123"))    # False
print(match_start_a_end_b("123b"))    # False
```

## 5. 删除 IP 地址中各段数字的前导零；全 0 保留一个 0

**思路**：遍历每一段数字，全 0 段直接返回 0；有有效数字时，去掉前面多余 0。

```python
def ip_remove_leading_zero(ip: str) -> str:
    def sub_func(part):
        s = part.group()
        if re.fullmatch(r'0+', s):
            return '0'
        return re.sub(r'0*([1-9]\d*)', r'\1', s)
    return re.sub(r'\d+', sub_func, ip)

# 测试用例
print(ip_remove_leading_zero("192.168.001.010"))  # 192.168.1.10
print(ip_remove_leading_zero("000.010.001.000"))  # 0.10.1.0
print(ip_remove_leading_zero("001.002.003.004"))  # 1.2.3.4
```

## 6. 将字符串中的空格替换为下划线，并将下划线替换为空格

**思路**：先把空格替换成临时标记，再把下划线换成空格，最后标记换回下划线，防止互相覆盖。

```python
def swap_space_underline(s: str) -> str:
    # 先用普通字符串replace，更简单稳定，不会报转义错误
    s1 = s.replace(' ', '\x01')
    s2 = s1.replace('_', ' ')
    s3 = s2.replace('\x01', '_')
    return s3

# 测试用例
print(swap_space_underline("hello world_test"))    # hello_world test
print(swap_space_underline("a b_c d"))             # a_b c_d
```

## 7. yyyy-mm-dd 格式的日期转换为 dd-mm-yyyy 格式

**思路**：分组捕获年、月、日，替换时调整分组顺序。

```python
def date_convert(s: str) -> str:
    pattern = r'^(\d{4})-(\d{2})-(\d{2})$'
    return re.sub(pattern, r'\3-\2-\1', s)

# 测试用例
print(date_convert("2026-11-24"))  # 24-11-2026
print(date_convert("2000-05-09"))  # 09-05-2000
```

## 8. 找出给定字符串中所有以 a 或 e 开头的单词

**思路**：`\b`单词边界，`[ae]`匹配 a/e 开头，`\w*`匹配后续单词字符。

```python
def find_word_start_ae(s: str) -> list:
    pattern = r'\b[ae]\w*'
    return re.findall(pattern, s)

# 测试用例
print(find_word_start_ae("apple cat egg ant book elephant"))
# ['apple', 'egg', 'ant', 'elephant']
print(find_word_start_ae("dog cat bird"))
# []
```

## 9. 将给定字符串中的 Road 缩写为 Rd

**思路**：单词边界匹配 Road，避免匹配单词内部，替换为 Rd。

```python
def road_to_rd(s: str) -> str:
    pattern = r'\bRoad\b'
    return re.sub(pattern, 'Rd', s)

# 测试用例
print(road_to_rd("Main Road, West Road"))    # Main Rd, West Rd
print(road_to_rd("Roadside Road"))            # Roadside Rd
```

## 10. 找出字符串中所有长度至少为 4 个字符的单词

**思路**：`\w{4,}`匹配 4 个及以上单词字符，搭配单词边界。

```python
def find_long_word(s: str) -> list:
    pattern = r'\b\w{4,}\b'
    return re.findall(pattern, s)

# 测试用例
print(find_long_word("hi hello test a ok python"))
# ['hello', 'test', 'python']
print(find_long_word("cat dog big"))
# []
```

## 11. 删除字符串中除字母和数字之外的所有字符

**思路**：`[^a-zA-Z0-9]`匹配非字母数字字符，全部替换为空。

```python
def keep_alnum_only(s: str) -> str:
    pattern = r'[^a-zA-Z0-9]'
    return re.sub(pattern, '', s)

# 测试用例
print(keep_alnum_only("he!llo@123#"))     # hello123
print(keep_alnum_only("A*B-3&9_xyz"))     # AB39xyz
```

## 12. 删除字符串中的括号及括号内的内容

**思路**：`\(.*?\)`非贪婪匹配括号和里面内容，替换为空，最后 strip 清理多余空格。

```python
def remove_bracket_content(s: str) -> str:
    pattern = r'\(.*?\)'
    return re.sub(pattern, '', s).strip()

# 测试用例
print(remove_bracket_content("example (.com) w3resource github (.com) stackoverflow (.com)"))
# example  w3resource github  stackoverflow
print(remove_bracket_content("test (abc) end"))
# test  end
```

## 13. BeautifulSoup 城市气候数据表

### 实际使用的 URL 和访问日期

- 主测试页面：Madison, Wisconsin
  - URL：<https://en.wikipedia.org/wiki/Madison,_Wisconsin>
- 额外测试页面：
  - Los Angeles：<https://en.wikipedia.org/wiki/Los_Angeles>
  - Hyderabad：<https://en.wikipedia.org/wiki/Hyderabad>
  - Boston（结构差异测试）：<https://en.wikipedia.org/wiki/Boston>
- 访问日期：**2026-11-24**
- 访问结果：以上页面均返回 HTTP 200，本次没有遇到 403、429 或其他 HTTP 错误。

网页内容和表格结构可能随更新时间变化，下面的结构说明和测试结果对应上述访问日期。

```python
# 如果 PyCharm 当前解释器缺少依赖，取消下一行注释并运行一次
# %pip install requests beautifulsoup4

import re
import socket
from functools import lru_cache
from urllib.parse import urlparse

import requests
from bs4 import BeautifulSoup, Tag

REQUEST_TIMEOUT_SECONDS = 10
REQUEST_HEADERS = {
    "User-Agent": "ClimateTableCourseDemo/1.0 (educational use)"
}
CLIMATE_TITLE_RE = re.compile(r"(?:^|\s)Climate\s+data\s+for\b", re.IGNORECASE)
SOURCE_ROW_RE = re.compile(r"^(?:sources?|data\s+sources?)\b", re.IGNORECASE)

MADISON_URL = "https://en.wikipedia.org/wiki/Madison,_Wisconsin"
LOS_ANGELES_URL = "https://en.wikipedia.org/wiki/Los_Angeles"
HYDERABAD_URL = "https://en.wikipedia.org/wiki/Hyderabad"
BOSTON_URL = "https://en.wikipedia.org/wiki/Boston"
NO_CLIMATE_TABLE_URL = "https://en.wikipedia.org/wiki/Python_(programming_language)"


def get_response(url: str) -> requests.Response:
    """获取响应；Wikipedia 域名连接失败时通过公共 DNS 重试一次。"""
    try:
        return requests.get(
            url,
            headers=REQUEST_HEADERS,
            timeout=REQUEST_TIMEOUT_SECONDS,
        )
    except requests.ConnectionError as original_error:
        host = urlparse(url).hostname or ""
        if not host.endswith("wikipedia.org"):
            raise

        # 某些网络环境的本地 DNS 可能返回错误地址，这里查询公共 DNS。
        try:
            dns_response = requests.get(
                "https://dns.google/resolve",
                params={"name": host, "type": "A"},
                headers=REQUEST_HEADERS,
                timeout=REQUEST_TIMEOUT_SECONDS,
            )
            dns_response.raise_for_status()
            ipv4 = next(
                item["data"]
                for item in dns_response.json().get("Answer", [])
                if item.get("type") == 1
            )
        except Exception as dns_error:
            raise original_error from dns_error

        original_getaddrinfo = socket.getaddrinfo

        def resolved_getaddrinfo(address, port, *args, **kwargs):
            if address == host:
                address = ipv4
            return original_getaddrinfo(address, port, *args, **kwargs)

        socket.getaddrinfo = resolved_getaddrinfo
        try:
            return requests.get(
                url,
                headers=REQUEST_HEADERS,
                timeout=REQUEST_TIMEOUT_SECONDS,
            )
        finally:
            socket.getaddrinfo = original_getaddrinfo


@lru_cache(maxsize=32)
def fetch_soup(url: str) -> BeautifulSoup:
    """读取网页并返回 html.parser 解析后的 BeautifulSoup 对象。"""
    response = get_response(url)
    response.raise_for_status()
    return BeautifulSoup(
        response.content,
        "html.parser",
        from_encoding=response.encoding or "utf-8",
    )


print("BeautifulSoup 气候表实验已准备，访问日期：2026-11-24")
```

### 任务 1：读取网页并观察气候表结构

状态；再用标准解析器 `html.parser` 创建 BeautifulSoup 对象。气候表标题不一定位于
`<caption>` 中，因此同时检查表格的 `<caption>` 和 `<th>` 文本。找到表格后，输出其
直接子标签、`tbody` 的直接子标签和行数。

```python
madison_soup = fetch_soup(MADISON_URL)
madison_table = None

for table in madison_soup.find_all("table"):
    title_texts = []
    if table.caption is not None:
        title_texts.append(table.caption.get_text(" ", strip=True))
    title_texts.extend(
        th.get_text(" ", strip=True) for th in table.find_all("th")
    )
    normalized_titles = [" ".join(text.split()) for text in title_texts]
    if any("climate data for" in text.lower() for text in normalized_titles):
        madison_table = table
        break

assert isinstance(madison_table, Tag)
table_children = [child.name for child in madison_table.children if child.name]
tbody = madison_table.find("tbody")
tbody_children = [child.name for child in tbody.children if child.name]
rows = madison_table.find_all("tr")

print("URL：", MADISON_URL)
print("返回类型：", type(madison_table).__name__)
print("表格类名：", madison_table.get("class"))
print("table 直接子标签：", table_children)
print("tbody 直接子标签：", tbody_children)
print("tbody 直接 tr 数量：", len(tbody_children))
print("表格 tr 总数：", len(rows))
print("标题行文本：", " ".join(rows[0].get_text(" ", strip=True).split())[:100], "...")
```

每行以行名 `<th>` 开头、后接数据 `<td>`；最后一行是天气数据来源。

### 任务 2：实现 `retrieve_climate_table(s)`

**思路**：函数只接收 URL 字符串。网络错误或 HTTP 错误必须抛出异常，因此使用
`raise_for_status()`；成功读取但找不到气候表才返回 `None`。标题识别同时支持表格
内的 `<caption>`、`<th>` 以及位于表格附近的情况。附近标题只有在表格同时具有
`Month`、`Jan`、`Feb` 等列时才被接受，以降低把邻近普通表格误判为气候表的风险。
页面有多个气候表时返回找到的第一个。

另外，`get_response()` 正常情况下直接调用 `requests.get()`；若本地 DNS 导致
Wikipedia 连接失败，会通过公共 DNS 重试一次。若重试仍失败，网络异常仍会抛出，
不会被误判为“没有气候表”。

```python
def normalize_text(text: str) -> str:
    """合并连续空白并去除首尾空白。"""
    return " ".join(text.split())


def cell_text(cell: Tag) -> str:
    """从表格单元格中取得规范化后的文本。"""
    return normalize_text(cell.get_text(" ", strip=True))


def has_climate_title(table: Tag) -> bool:
    """检查表格自身的 caption 或 th 是否包含气候表标题。"""
    if table.caption is not None:
        caption_text = normalize_text(table.caption.get_text(" ", strip=True))
        if CLIMATE_TITLE_RE.search(caption_text):
            return True

    return any(
        CLIMATE_TITLE_RE.search(cell_text(header))
        for header in table.find_all("th")
    )


def looks_like_monthly_climate_table(table: Tag) -> bool:
    """检查表格是否具有 Month、Jan、Feb 月份列。"""
    for row in table.find_all("tr")[:4]:
        texts = {
            cell_text(cell).casefold()
            for cell in row.find_all(["th", "td"], recursive=False)
        }
        if "month" in texts and {"jan", "feb"}.issubset(texts):
            return True
    return False


def has_nearby_climate_title(table: Tag) -> bool:
    """处理标题位于表格附近而不是表格内部的情况。"""
    for node in table.find_all_previous(
        ["caption", "h2", "h3", "h4", "p", "th"],
        limit=8,
    ):
        text = normalize_text(node.get_text(" ", strip=True))
        if CLIMATE_TITLE_RE.search(text):
            return looks_like_monthly_climate_table(table)
    return False


def is_climate_table(table: Tag) -> bool:
    """判断一个 table 节点是否为气候数据表。"""
    return has_climate_title(table) or has_nearby_climate_title(table)


def retrieve_climate_table(s: str) -> Tag | None:
    """返回城市网页中的气候数据表；成功访问但无表时返回 None。"""
    soup = fetch_soup(s)
    for table in soup.find_all("table"):
        if is_climate_table(table):
            return table
    return None


print("retrieve_climate_table(s) 已定义")
```

```python
city_urls = {
    "Madison": MADISON_URL,
    "Los Angeles": LOS_ANGELES_URL,
    "Hyderabad": HYDERABAD_URL,
    "Boston": BOSTON_URL,
}

for city, url in city_urls.items():
    table = retrieve_climate_table(url)
    assert isinstance(table, Tag)
    assert table.name == "table"
    print(f"{city}: 返回 Tag，表格类名={table.get('class')}")

# 成功读取页面但没有气候表时返回 None
assert retrieve_climate_table(NO_CLIMATE_TABLE_URL) is None
print("无气候表页面：正确返回 None")

# HTTP 错误必须抛出异常，而不是返回 None
bad_url = (
    "https://en.wikipedia.org/wiki/"
    "This_page_should_not_exist_climate_demo_1048576"
)
try:
    retrieve_climate_table(bad_url)
except requests.HTTPError as error:
    print("HTTP 错误：正确抛出异常，状态码 =", error.response.status_code)
else:
    raise AssertionError("不存在的页面应抛出 HTTPError")
```

### 任务 3：实现 `list_climate_table_row_names(s)`

**思路**：复用 `retrieve_climate_table(s)` 定位表格；成功但找不到表格时返回 `None`。
遍历每个 `<tr>`，只检查直接子单元格；气候数据行应同时具有行标题 `<th>` 和数据
单元格 `<td>`。如果存在 `scope="row"` 就优先使用它，否则使用行内第一个 `<th>`，
因此也能处理 Boston 页面中缺少 `scope` 的表格。最后排除气候表标题行、`Month`
列标题和来源行，并保证输出是去除首尾空白的 Python 字符串。

```python
def list_row_names_from_table(table: Tag) -> list[str]:
    """从一个已定位的气候表 Tag 中提取行名。"""
    row_names = []

    for row in table.find_all("tr"):
        cells = row.find_all(["th", "td"], recursive=False)
        headers = [cell for cell in cells if cell.name == "th"]
        data_cells = [cell for cell in cells if cell.name == "td"]

        # 表头行没有数据单元格；来源行通常没有行标题单元格。
        if not headers or not data_cells:
            continue

        # 排除包含 Climate data for 的表格标题行。
        if any(CLIMATE_TITLE_RE.search(cell_text(header)) for header in headers):
            continue

        # scope="row" 可用时优先使用；缺失时退回到第一个 th。
        row_header = next(
            (
                header
                for header in headers
                if header.get("scope", "").casefold() == "row"
            ),
            headers[0],
        )
        name = cell_text(row_header)

        if not name:
            continue
        if name.casefold() == "month":
            continue
        if SOURCE_ROW_RE.match(name):
            continue

        row_names.append(name)

    return row_names


def list_climate_table_row_names(s: str) -> list[str] | None:
    """返回气候表行名；成功读取页面但无气候表时返回 None。"""
    table = retrieve_climate_table(s)
    if table is None:
        return None
    return list_row_names_from_table(table)


print("list_climate_table_row_names(s) 已定义")
```

```python
madison_row_names = list_climate_table_row_names(MADISON_URL)
assert madison_row_names is not None
assert all(
    isinstance(name, str) and name == name.strip()
    for name in madison_row_names
)
assert "Month" not in madison_row_names
assert not madison_row_names[0].startswith("Climate data for")
assert not madison_row_names[-1].startswith("Source")

print(f"Madison 行名数量：{len(madison_row_names)}")
for index, name in enumerate(madison_row_names, start=1):
    print(f"{index:02d}. {name}")

# 成功读取但没有气候表时返回 None
assert list_climate_table_row_names(NO_CLIMATE_TABLE_URL) is None
print("无气候表页面：行名函数正确返回 None")

# 额外检查：行标题没有 scope 属性时仍能正确提取。
no_scope_html = '''
<table>
  <tr><th colspan="3">Climate data for Example</th></tr>
  <tr><th>Month</th><th>Jan</th><th>Feb</th></tr>
  <tr><th>Record high °C (°F)</th><td>30</td><td>31</td></tr>
</table>
'''
no_scope_table = BeautifulSoup(no_scope_html, "html.parser").table
assert list_row_names_from_table(no_scope_table) == ["Record high °C (°F)"]
print("无 scope 属性结构：行名提取正确")
```

**完成情况**：`retrieve_climate_table(s)` 已覆盖有气候表、无气候表和 HTTP 错误三种
情况；`list_climate_table_row_names(s)` 已排除标题、月份及来源行，并兼容有无
`scope="row"` 的表格结构。
